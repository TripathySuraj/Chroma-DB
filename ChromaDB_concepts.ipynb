#**2. Create a Chroma Client -> Chroma DB in-memory client**
* ***It starts a Chroma server in-memory, so any data you ingest will be lost when your program terminates***

```python
import chromadb
chroma_client = chromadb.Client()
```

#**3. Create a collection**
* Collections are where you’ll store your embeddings, documents, and any additional metadata. Collections index your embeddings and documents, and enable efficient retrieval and filtering.
* **collection** is your Python variable, whereas "**my_collection**" is the actual name of the collection stored inside ChromaDB

```python
collection = chroma_client.get_or_create_collection(name="my_collection")
```

#**4. Add some text documents to the collection**

#**5. Query the collection**
* **query_texts** is not itself a filter. It is the natural-language input used for similarity search. The **where** and **where_document** parameters restrict which documents can be returned.

* Note: **contains** and **not_contains** perform text substring matching, not semantic matching. Numerical operators work on numeric metadata values, so store years, prices and ratings as numbers rather than strings.

```python
from pprint import pprint

results = collection.query(
    query_texts=["This is a query document about hawaii"], # Chroma will embed this for you
    n_results=4 # how many results to return
)

pprint(results)
```

#**6. Filtering Documents based on metadata and document conditions**

```python
results = collection.get(
    where={"category": {"$eq": "fruit"}}
)

pprint(results["documents"])
```

```python
results = collection.get(
    where={"year": {"$gte": 2024}}
)

pprint(results["documents"])
```

```python
results = collection.get(
    where={
        "$and": [
            {"category": "fruit"},
            {"price": {"$gt": 50}}
        ]
    }
)

pprint(results["documents"])
```

```python
results = collection.get(
    where_document={
        "$or": [
            {"$contains": "apples"},
            {"$contains": "oranges"}
        ]
    }
)

print(results["documents"])
```

```python
results = collection.query(
    query_texts=["Explain artificial intelligence and programming"],
    where={
        "$and": [
            {"category": "technology"},
            {"price": {"$gt": 300}}
        ]
    },
    n_results=3
)

pprint(results['documents'])
```

#**7. Using custom embeddings**
* **Using huggung face embedding model -> Qwen/Qwen3-VL-Embedding-2B**

**a. Using Persistent client**
* ***You can configure Chroma to save and load the database from your local machine, using the PersistentClient.Data will be persisted automatically and loaded on start (if it exists).***

```python
client = chromadb.PersistentClient(
    path="/content/chroma_db"
)
```

**b. Installing Sentence-Transformers**

```python
import subprocess
import sys

# Install sentence-transformers to easily load Hugging Face embedding models
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "sentence-transformers"])
```

**c. Loading embedding model**

```python
from chromadb.utils import embedding_functions

# Initialize the Hugging Face embedding function using sentence-transformers
# We will use the model specified or fall back to its sentence-transformer representation
embedding_qwen = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="Qwen/Qwen3-VL-Embedding-2B",
    device="cuda"
)
```

**d. Making collection using chromaDb client and custom embedding model**

```python
collection = client.get_or_create_collection(
    name="employee",
    embedding_function=embedding_qwen
)
```
